```python
# 3. Create the following DataFrame:

# | Name  | Age | Department | Marks |
# | ----- | --: | ---------- | ----: |
# | Asha  |  22 | CSE        |    85 |
# | Ravi  |  23 | ECE        |    78 |
# | Meena |  21 | CSE        |    92 |
# | Kiran |  24 | ME         |    74 |
# | Arun  |  22 | ECE        |    88 |

# Display:

# 	Complete DataFrame
df = pd.DataFrame({
    'Name': ['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
    'Age': [22, 23, 21, 24, 22],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Marks': [85, 78, 92, 74, 88]
})
print("Complete DataFrame: \n")
print(df)
# 	First 3 rows
print("\nFirst 3 rows: \n")
print(df.head(3))

# 	Last 2 rows
print("\nLast 2 rows: \n")
print(df.tail(2))

# 	Shape
print("\nShape: ", df.shape)

# Column names
print("\n\nColumn names:", df.columns.tolist())

# Data types
print("\nData types:")
print(df.dtypes)
```

```python
# 7. Find:

# 	Students scoring above 80.
print("Scoring above 80:\n", df[df['Marks'] > 80])

# 	Students scoring between 75 and 90.
print("\nBetween 75 and 90:\n", df[(df['Marks'] >= 75) & (df['Marks'] <= 90)])

# 	Students belonging to CSE.
print("\nCSE students:\n", df[df['Department'] == 'CSE'])


# 	CSE students scoring above 80.
print("\nCSE students scoring above 80:\n",
      df[(df['Department'] == 'CSE') & (df['Marks'] > 80)])
```

```python
# 8. Display students belonging to either CSE or ECE.

print(df[df['Department'].isin(['CSE', 'ECE'])])
```

```python
# 9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
import numpy as np
df['Result'] = np.where(df['Marks'] >= 75, 'Pass', 'Fail')
print(df)
```

```python
# 10. Assign grades according to:

# 	| Marks | Grade |
# 	| ----: | ----- |
# 	|  ≥ 90 | A     |
# 	|  ≥ 80 | B     |
# 	|  ≥ 70 | C     |
# 	|  < 70 | D     |

def assign_grade(m):
    if m >= 90:
        return 'A'
    elif m >= 80:
        return 'B'
    elif m >= 70:
        return 'C'
    else:
        return 'D'
 
df['Grade'] = df['Marks'].apply(assign_grade)
print(df)
```

```python
# 12. Consider:

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

# Find:

# Missing values
print("Missing values (boolean mask):\n", students.isnull())

# Number of missing values in each column
print("\nNumber of missing values per column:\n", students.isnull().sum())

# Rows containing missing values
print("\nRows containing missing values:\n", students[students.isnull().any(axis=1)])
```

```python
# 13. Replace:

# 	Missing Age with average Age
students_filled = students.copy()
students_filled['Age'] = students_filled['Age'].fillna(students_filled['Age'].mean())


# 	Missing Marks with average Marks
students_filled['Marks'] = students_filled['Marks'].fillna(students_filled['Marks'].mean())
print(students_filled)
```

```python
# 14. Remove rows containing missing values.
students_dropped = students.dropna()
print(students_dropped)
```

```python
# 17. Consider:

data = {
    "Name": [
        "Asha", "Ravi", "Meena","Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE","ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df17 = pd.DataFrame(data)


# Find:

# Average marks by department
print("Average marks by department:\n", df17.groupby('Department')['Marks'].mean())
# Maximum marks by department
print("\nMaximum marks by department:\n", df17.groupby('Department')['Marks'].max())

# Number of students in each department
print("\nNumber of students per department:\n", df17.groupby('Department')['Name'].count())

# For each department, calculate:
# Mean
# Minimum
# Maximum
# Count
print("\nMean, Min, Max, Count per department:\n",
      df17.groupby('Department')['Marks'].agg(['mean', 'min', 'max', 'count']))
```

```python
# 19. Use the following dataset:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)


# Perform the following analysis:

# Display the first five records.

print("First five records:\n", df.head())
# Find the dimensions of the dataset.
print("\nDimensions:", df.shape)

# Find the average final mark.
print("\nAverage final mark:", df['FinalMarks'].mean())

# Find the highest-scoring student.
print("\nHighest-scoring student:\n", df.loc[df['FinalMarks'].idxmax()])

# Find the lowest-scoring student.
print("\nLowest-scoring student:\n", df.loc[df['FinalMarks'].idxmin()])

# Display students scoring above 80.
print("\nStudents scoring above 80 (FinalMarks):\n", df[df['FinalMarks'] > 80])

# Find average marks by department.
print("\nAverage marks by department:\n", df.groupby('Department')['FinalMarks'].mean())

# Find students having attendance above 80%.
print("\nAttendance above 80%:\n", df[df['Attendance'] > 80])
# Sort students by FinalMarks.
print("\nSorted by FinalMarks:\n", df.sort_values('FinalMarks'))
# Add a Result column: Pass if FinalMarks ≥ 70, otherwise Fail.

# Add a performance category:
# 	≥90 → Excellent
# 	≥80 → Very Good
# 	≥70 → Good
# 	<70 → Needs Improvement


# Find the department having the highest average FinalMarks. 
df['Result'] = np.where(df['FinalMarks'] >= 70, 'Pass', 'Fail')
 
def performance_category(m):
    if m >= 90:
        return 'Excellent'
    elif m >= 80:
        return 'Very Good'
    elif m >= 70:
        return 'Good'
    else:
        return 'Needs Improvement'
 
df['Performance'] = df['FinalMarks'].apply(performance_category)
print("\nWith Result and Performance columns:\n", df)
 
dept_avg = df.groupby('Department')['FinalMarks'].mean()
print("\nDepartment with highest average FinalMarks:",
      dept_avg.idxmax(), "->", dept_avg.max())
```
