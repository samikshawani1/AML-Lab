### 2(d) Principal Component Analysis (PCA)

**Aim:** To implement PCA and visualize high-dimensional data using a reduced number of dimensions.

**Theory:**

Principal Component Analysis is an unsupervised dimensionality reduction technique. It transforms the original features into a smaller set of uncorrelated variables called principal components.

The first principal component captures the maximum possible variance, while subsequent components capture as much of the remaining variance as possible.

PCA is commonly used for visualization, data compression, and reducing feature dimensionality.

**Dataset:** Iris dataset available in Scikit-learn.

**Evaluation:** Explained variance ratio.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

iris = load_iris()
X, y = iris.data, iris.target
X_scaled = StandardScaler().fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_.round(4))
print("Total variance explained by 2 components:", round(pca.explained_variance_ratio_.sum(), 4))
```

```python
# Graph: samples in the first two principal components
plt.figure(figsize=(7, 5))
for label, name in enumerate(iris.target_names):
    plt.scatter(X_pca[y == label, 0], X_pca[y == label, 1], label=name, alpha=0.8)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection of Iris Dataset")
plt.legend()
plt.tight_layout()
plt.show()

```

```python
# Graph: variance explained by each principal component
pca_all = PCA().fit(X_scaled)
plt.figure(figsize=(7, 4))
plt.bar(range(1, len(pca_all.explained_variance_ratio_) + 1),
        pca_all.explained_variance_ratio_)
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("PCA Explained Variance")
plt.xticks(range(1, len(pca_all.explained_variance_ratio_) + 1))
plt.tight_layout()
plt.show()
```

**Result and Conclusion:**

PCA was applied to the standardized Iris dataset to reduce its four original features to two principal components.

The reduced dataset was visualized in a two-dimensional scatter plot. An explained variance graph was also generated to examine how much variance is captured by the principal components.

PCA provides a useful way to visualize high-dimensional data while retaining a substantial proportion of its variance.
