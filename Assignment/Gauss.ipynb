### 2(c) Implementation of Gaussian Mixture Model (GMM)

**Aim:** To implement Gaussian Mixture Models for clustering and visualize the resulting clusters.

**Theory:**

A Gaussian Mixture Model is a probabilistic clustering algorithm that assumes the dataset is generated from a mixture of multiple Gaussian distributions.

Unlike K-means, GMM assigns probabilities to cluster membership and can model clusters with different shapes and orientations.

- **AIC:** Akaike Information Criterion, used to compare statistical models.
- **BIC:** Bayesian Information Criterion, which penalizes model complexity.

Lower AIC and BIC values are generally preferred when comparing models fitted to the same dataset.

**Evaluation Metric:** Silhouette score.

**Dataset:** Generated sample data containing three clusters.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

X, _ = make_blobs(n_samples=250, centers=3, cluster_std=1.1, random_state=42)
gmm = GaussianMixture(n_components=3, random_state=42)
labels = gmm.fit_predict(X)
print("GMM silhouette score:", round(silhouette_score(X, labels), 4))
print("AIC (lower is better when comparing same dataset):", round(gmm.aic(X), 2))
print("BIC (lower is better when comparing same dataset):", round(gmm.bic(X), 2))
```

```python
# Graph: GMM cluster assignments and Gaussian component centers
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, alpha=0.75)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], marker="X", s=180,
            edgecolor="black", label="Component means")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Gaussian Mixture Model Clusters")
plt.legend()
plt.tight_layout()
plt.show()
```

**Result and Conclusion:**

The Gaussian Mixture Model was implemented with three components. The resulting clusters and component means were visualized.

The model was evaluated using the silhouette score, AIC, and BIC. These measures help analyze cluster separation and compare GMM configurations.
