```python
import pandas as ps
import numpy as nm
import matplotlib.pyplot as plt
```

```python
df = ps.read_csv("D:\BDA 20\AIML-BDA-Lab-Work\data\Social_Network_Ads.csv")
df.head()
```

```python
print("InDependent variable, X:")
x = df.iloc[:, [2, 3]].values                 #[:, :-1] --- all column; not last one
print(x)

print("Dependent variable, Y:")
y = df.iloc[:, 4].values
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y,
    test_size=0.25,
    random_state=45
)
```

```python
from sklearn.preprocessing import StandardScaler
sc_x = StandardScaler()
x_train = sc_x.fit_transform(x_train) 
x_test=sc_x.transform(x_test)

print(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
#minkowski = Eudclidean distance
classifier = KNeighborsClassifier(n_neighbors= 5, metric= 'minkowski', p = 2)
classifier.fit(x_train, y_train)
```

```python
# Prediciting the TEST set results

y_pred = classifier.predict(x_test)         #here we give X_testing data and prediciting Y data

print("Actual: ",y_test)
print("Predicted: ",y_pred)
```

```python
#Making confusion matrix

from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)

print("True Pos", cm[0][0])
print("True Neg", cm[0][1])
print("False Pos", cm[1][0])
print("False Neg", cm[1][1])
```

```python
#finding Accuracy of matrix

accuracy = (cm[0][0]  + cm[1][1]) / (cm[0][0]  + cm[1][1] + cm[1][0] + cm[0][1])
print("Accuracy - ", accuracy)
```

```python
#find accuracy with inbuilt 

from sklearn import metrics
predicition = metrics.accuracy_score(y_test, y_pred)

print("Accuracy score: ", predicition * 100, "%")
```

```python
#it return complete report of it 

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
new = nm.array([[29,80000]])

new_scl = sc_x.transform(new)

predicition = classifier.predict(new_scl)

if predicition[0] == 1:
    print("Predicted to Purchased")
else:
    print("Not Predicted to Purchased")
```

<h1>K - Negibour 

```python
import math
from collections import Counter

def euclidean_dist(p1, p2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))

def knn(train_data, train_labels, test_point, k):
    distances = []

    for i in range(len(train_data)):
        dist = euclidean_dist(train_data[i], test_point)
        distances.append((dist, train_labels[i]))

    distances.sort(key=lambda x: x[0])
    k_nearest = distances[:k]

    labels = [label for _, label in k_nearest]
    return Counter(labels).most_common(1)[0][0]
```

```python
x_train = [(2,4), (4,4), (4,6), (6,8)]
y_train = ["Red", "Red", "Blue", "Blue"]

test_point = (5,5)
k = 3

print(knn(x_train, y_train, test_point, k))
```
